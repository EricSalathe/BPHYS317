## Dynamics

Newton's second law for each cart is

$$
f_i=m_i\ddot{x}_i.
$$

With $m_i=1$ and $\vec f=-K\vec x$,

$$
-K\vec x=\frac{d^2}{dt^2}\vec x.
$$

Assume a solution of the form

$$
\vec x(t)=\vec u e^{i\omega t},
$$

where $\vec u$ is a constant amplitude vector. Substitution gives

$$
K\vec u=\omega^2\vec u.
$$

This is the eigenvalue problem

$$
K\vec u=\lambda\vec u,
\qquad
\lambda=\omega^2.
$$

### 5. Two-mass system

For the two-mass system,

$$
K=\begin{bmatrix}2&-1\\-1&2\end{bmatrix}.
$$

Find the possible eigenvectors $\vec u$ and angular frequencies $\omega$.

````python
K2 = np.array([[2.0, -1.0],
               [-1.0, 2.0]])

eigenvalues2, eigenvectors2 = np.linalg.eigh(K2)
omega2 = np.sqrt(eigenvalues2)

print('eigenvalues =', eigenvalues2)
print('angular frequencies =', omega2)
print('eigenvectors (columns) =\n', eigenvectors2)
````
