Here's sample code to get started. Just change M.

```python
from numpy import array, sin, cos, arccos, arcsin, pi, diag
from numpy.linalg import eig

# Sample code for eigenvalues and eigenvectors
M = array([[0, 2],[ 1, -1]])
d, U = eig(M)

D = diag(d)

display(Math(r"M ="+latex(Matrix(M))))
display(Math(r"D ="+latex(Matrix(D))))
display(Math(r"U ="+latex(Matrix(U))))

```
