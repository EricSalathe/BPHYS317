1)  For the Pauli Spin Matrices: 



show:
a) $A^2 = B^2 = C^2 $= a unit matrix

b) AB=-BA, BC=-CB, AC=-CA (that is, the matrices anticommute)

c) AB-BA=2iC (AB-BA is called the commutator of A and B)



```python
A = array([[0,1],[1,0]])
B = array([[0,-i],[i,0]])
C = array([[1,0],[0,-1]])

print("a)")
display(Math(r"A^2 ="+latex(Matrix(matmul(A, A)))))
display(Math(r"B^2 ="+latex(Matrix(matmul(B, B)))))
display(Math(r"C^2 ="+latex(Matrix(matmul(C, C)))))

print("b)")
display(Math(r"AB ="+latex(Matrix(matmul(A, B)))))
display(Math(r"BA ="+latex(Matrix(matmul(B, A)))))

print("c)")
ABBA = matmul(A, B) - matmul(B, A)
display(Math(r"AB-BA ="+latex(Matrix(ABBA))))
```

6) Find $e^{i\theta A}$ for the Pauli spin matrix,
$$
A=
  \left[ {\begin{array}{cc}
   0 & 1 \\
   1 & 0 \\
  \end{array} } \right] 
$$
Hint: Use the power series expansion of $e^x$, $
  e^x=1\:+\frac{\:x}{1!}+\frac{x^2}{2!}+\frac{x^3}{3!}+\:...$

Using the Taylor series for the exponential,

$$
e^{i\theta A} = I + \frac{1}{1!}\,i\theta A + \frac{1}{2!}\,(i\theta)^2 A^2 + \frac{1}{3!}\,(i\theta)^3 A^3 + \cdots
$$

From #1, $A^2 = I$, so $A^3 = A^2 A = A$. By extension,

$$
A^n =
\begin{cases}
I, & n \text{ even},\\
A, & n \text{ odd}.
\end{cases}
$$

Likewise, $i^n = \pm 1$ for $n$ even and $i^n = \pm i$ for $n$ odd. Substituting,

$$
e^{i\theta A} = I + i\theta A - \frac{\theta^2}{2!}\,I - i\,\frac{\theta^3}{3!}\,A + \frac{\theta^4}{4!}\,I + \cdots
$$

Grouping the terms with $I$ and the terms with $A$,

$$
e^{i\theta A} = I\left(1 - \frac{\theta^2}{2!} + \frac{\theta^4}{4!} - \cdots\right)
              + iA\left(\theta - \frac{\theta^3}{3!} + \frac{\theta^5}{5!} - \cdots\right)
$$

The expressions in parentheses are the Taylor series of $\cos\theta$ and $\sin\theta$, respectively, so

$$
e^{i\theta A} = I\cos\theta + iA\sin\theta
$$
