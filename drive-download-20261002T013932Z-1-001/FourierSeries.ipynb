```python
from numpy.linalg import eig, inv, det
from numpy import shape, array, matmul
```

Define a matrix and find its eigenvectors and eigenvalues

```python
A = array([[ 0.8, 0.3],[ 0.2, 0.7]])
w, U = eig(A)

print("Eigen values w=",w)
print("Eigen vectors:")
print("u1 = ", U[:,0])
print("u2 = ", U[:,1])
```
