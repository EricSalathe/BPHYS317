1. a system with one side free (ie one wall, 2 masses, two springs), the K matrix is 

$K=
  \left[ {\begin{array}{cc}
   2 & -1 \\
   -1 & 1 \\
  \end{array} } \right]$

What are the eigenvalue and eigenvectors for this system? What are the frequencies the system oscillate at? Plot the 2-D eigenvectors.

```python
K = array([[2,-1],[-1,1]])
w, U = eig(K)

print("Eigen values w=",around(w,3))
print("Frequencies =", around(sqrt(w),3))
print("Eigen vectors:")
print("u1 = ", around(U[:,0],3))
print("u2 = ", around(U[:,1],3))
```

2 a.  Find the eigenvalues and eigenvectors of the following matrix
$M=
  \left[ {\begin{array}{cc}
   3&2&4\\
   2&0&2\\
  4&2&3
  \end{array} } \right]$

```python
M = array([[3,2,4],[2,0,2],[4,2,3]])
w, U = eig(M)

print("Eigen values w=",w)
print("Eigen vectors:")
print("u1 = ", around(U[:,0],3))
print("u2 = ", around(U[:,1],3))
print("u3 = ", around(U[:,2],3))
```

Note that the eigenvalue -1 is repeated. This means that the associated eigenvectors are not unique. $\vec{u}_1$ and $\vec{u}_1$ could be any two orthogonal vectors in the plane perpedicular to $\vec{u}_2$
