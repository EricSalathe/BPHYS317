Define a matrix and find its eigenvectors and eigenvalues

```python
A = array([[1, 3],[ 2, 2]])
w, U = eig(A)

print('w')
print(w)

print('U')
print(U)

print()
print("Eigen values w=",w)
print("Eigen vectors are columns of U:")
print("u1 = ", U[:,0])
print("u2 = ", U[:,1])
```
